```python
import numpy as np
import matplotlib.pyplot as plt

A = 5
B = 3.5
n = 10

xi = np.sort(np.random.randint(-B/A , 40, n))
xi_points = np.linspace(min(xi) , max(xi) , 1000)
def f(x):
  return x/(A + B * x)

plt.plot(xi_points, f(xi_points), label = "y = x/(A + Bx)", color = "pink")
plt.scatter(xi, f(xi), label = "points", color = "pink")

epsilon = 0.2 * min(f(xi))
noise = np.random.uniform(-epsilon , epsilon , 10)
#print(noise)


yi = []
for i in range(len(xi)):
  yi.append(f(xi[i]) + noise[i])


plt.scatter(xi, yi, label = "points with noise", color = "blue")
A_mat = []
for i in range(len(xi)):
  A_mat.append([yi[i], yi[i]*xi[i]])

#print(A_mat)
Q, R = np.linalg.qr(A_mat)

b = np.linalg.solve(R, Q.T @ xi)
print("Получившиеся значения коэффициентов:")
print("A =" , b[0], ", B =" , b[1])

def f_MNK(x):
  return x/(b[0] + b[1] * x )


plt.plot(xi_points, f_MNK(xi_points),"r--", label = "f_MNK" )
plt.grid()
plt.xlabel(" x ")
plt.ylabel(" f(x) ")
plt.xlim(-0.7, 40)
plt.legend()
```
